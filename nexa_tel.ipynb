```python
import pandas as pd

# Step 1: Ensure change_date is parsed correctly
plan_history['change_date'] = pd.to_datetime(
    plan_history['change_date'], format='%d-%m-%Y', errors='coerce'
)

# Step 2: Get latest new_plan_id per customer
latest_plan = (
    plan_history.sort_values('change_date')
    .groupby('customer_id')
    .tail(1)[['customer_id','new_plan_id']]
)

# Rename column to plan_id for consistency
latest_plan.rename(columns={'new_plan_id':'plan_id'}, inplace=True)

# Step 3: Merge with subscriptions
subscriptions = subscriptions.merge(
    latest_plan, on='customer_id', how='left', suffixes=('', '_hist')
)

# Step 4: Fill missing plan_id from latest_plan
subscriptions['plan_id'] = subscriptions['plan_id'].fillna(subscriptions['plan_id_hist'])
subscriptions.drop(columns=['plan_id_hist'], inplace=True)

# Step 5: Final check
missing_after = subscriptions['plan_id'].isnull().sum()
print("Remaining null plan_id after recovery:", missing_after)

# Optional: Drop rows if still null
subscriptions = subscriptions[subscriptions['plan_id'].notnull()]
```
